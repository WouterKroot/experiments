```python
import numpy as np
import matplotlib.pyplot as plt

def sigmoid(z, mu, slope):
    return 1 / (1 + np.exp(-slope * (z - mu)))


z = np.linspace(-0.9, -0.75, 100)
a = sigmoid(z, mu=-0.82, slope=100)
plt.plot(z, a)

x = np.arange(-0.9, -0.75, 0.01)
y = sigmoid(x, mu=-0.82, slope=100)
plt.plot(x, np.zeros(len(x)), 'ro')     # dots on x-axis
plt.plot(x, y, 'ro')

y_sig = sigmoid(x, mu=-0.82, slope=100)

                          # smooth curve
plt.vlines(x, 0, y_sig, colors='r', linestyles='dashed', alpha=0.5)  # vertical lines 
```
